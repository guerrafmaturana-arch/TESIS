```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar el dataset detectando el delimitador real (, o ;)
dataset = pd.read_csv('DATASET2.csv', sep=None, engine='python')

# 2. Si Cancha_CuIn NO está en el archivo, calcularla por balance de masas:
if 'Cancha_CuIn' not in dataset.columns:
    dataset['Cancha_CuIn'] = dataset['Cancha_CuT'] - dataset['Cancha_CuS']

# 3. Ordenar las columnas disponibles de forma consistente
columnas_deseadas = ['Cancha_Ton', 'Cancha_CuT', 'Cancha_CuS', 'Cancha_CuIn', 'Canje_CuT']
columnas_existentes = [col for col in columnas_deseadas if col in dataset.columns]
df_analisis = dataset[columnas_existentes]

# 4. Calcular la matriz de correlación de Pearson
corr_matrix = df_analisis.corr(method='pearson')

print("="*65)
print("             MATRIZ DE CORRELACIÓN DE PEARSON")
print("="*65)
print(corr_matrix.round(3))

# 5. Generar el Heatmap con formato idéntico
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".3f",
    vmin=-0.2,
    vmax=1.0,
    cbar_kws={'label': 'Coeficiente r de Pearson'}
)
plt.title("Matriz de Correlación de Pearson", fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```
